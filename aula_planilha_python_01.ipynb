```python
# Importando uma biblioteca para dados (Pandas)
import pandas as pd
```

```python
# Acessar a minha planilha em um repositório externo.
# URL é uma variável.
URL = "https://raw.githubusercontent.com/wuldson-franco/T01-amazing-data-ia-lab/main/base/Cafe_Tambau_Vendas_Setembro.xlsx"
```

```python
# Ler a planilha.
# 1) abrir a planilha (o cabeçalho está na linha 4)
vendas = pd.read_excel(URL, sheet_name="Vendas Set", header=3)
```

```python
# Visualizar os registros da planilha.
vendas.head()
```

```python
# Função para verificar o quantitativo de linhas e colunas da planilha.
vendas.shape
```

```python
# Função para analisar o detalhamento das colunas da planilha.
vendas.info()
```

```python
# 2) tirar o total e o rodapé: só linhas com número de pedido.
vendas = vendas.dropna(subset=["Pedido"])
```

```python
# Apresentar o quantitativo de linhas preenchidas pós tratamento.
print(len(vendas))
```

```python
# Verificando a última linha do meu dataframe.
vendas.tail()
```

```python
# 3) SOMASE: faturamento por canal
por_canal = vendas.groupby("Canal")["Total (R$)"].sum()
```

```python
# 1. Carregar a aba de cadastro com os nomes dos produtos
produtos = pd.read_excel(URL, sheet_name="Produtos")

# 2. Somar a quantidade de vendas agrupando pelo código do produto
total_vendas = vendas.groupby("Cód. Produto")["Qtde."].sum().reset_index()

# 3. Juntar as duas tabelas para termos o Código, o Nome e a Quantidade
tabela_final = pd.merge(total_vendas, produtos, left_on="Cód. Produto", right_on="id_produto")

# 4. Localizar a linha do produto campeão (com a maior quantidade)
campeao = tabela_final.loc[tabela_final["Qtde."].idxmax()]

# 5. Mostrar o resultado na tela de forma simples
print(f"Produto campeão: {campeao['nome_produto']} ({campeao['Cód. Produto']})")
print(f"Total de unidades vendidas: {campeao['Qtde.']:.0f}")
```

### Explicação do Código:

1. **`pd.read_excel(..., sheet_name="Metas", skiprows=1)`**: Lemos a aba de metas do seu arquivo Excel pulando a primeira linha de título (`skiprows=1`). Isso permite que o Pandas alinhe corretamente as colunas de **Categoria** e **Meta (R$)** sem lixo ou valores vazios.
2. **`pd.merge(vendas, produtos, ...)`**: Associamos cada linha de venda à sua respectiva categoria cadastrada, garantindo consistência nos nomes das categorias.
3. **`groupby("categoria")["Total (R$)"].sum()`**: Agrupamos e somamos todo o faturamento real de setembro para cada uma das quatro categorias.
4. **`pd.merge(..., on="categoria")`**: Cruzamos dinamicamente os resultados de faturamento real com a tabela de metas importada do Excel.
5. **Comparação Direta**: Criamos o indicador lógico para verificar se o Faturamento foi maior ou igual à Meta oficial, gerando os emojis de status de forma automatizada.

```python
import pandas as pd

# 1. Carregar a aba de Metas oficial do Excel (pulando o cabeçalho decorativo)
metas_oficiais = pd.read_excel(URL, sheet_name="Metas", skiprows=1)
metas_oficiais.columns = ["categoria", "Meta (R$)"]
metas_oficiais = metas_oficiais.dropna(subset=["categoria"])

# Ajustar nomes de categorias para garantir que batam perfeitamente (ex: 'Salgado' vs 'Salgados')
# No cadastro temos 'Salgado' e 'Doce', e na aba de metas também. Vamos garantir a consistência:
produtos["categoria"] = produtos["categoria"].str.strip()
metas_oficiais["categoria"] = metas_oficiais["categoria"].str.strip()

# 2. Mesclar as vendas com o cadastro de produtos para obter a categoria correta
vendas_com_categoria = pd.merge(vendas, produtos, left_on="Cód. Produto", right_on="id_produto")

# 3. Calcular o faturamento real acumulado agrupando por categoria
faturamento_por_categoria = vendas_com_categoria.groupby("categoria")["Total (R$)"].sum().reset_index()

# 4. Cruzar o faturamento real com as metas oficiais obtidas diretamente da planilha
analise_metas = pd.merge(faturamento_por_categoria, metas_oficiais, on="categoria")

# 5. Criar a coluna de atingimento e o status visual formatado
analise_metas["Atingiu?"] = analise_metas["Total (R$)"] >= analise_metas["Meta (R$)"]
analise_metas["Status"] = analise_metas["Atingiu?"].map({True: "✅ Atingiu", False: "❌ Não Atingiu"})

# Exibir a tabela final consolidada com precisão absoluta
display(analise_metas[["categoria", "Total (R$)", "Meta (R$)", "Status"]])
```
